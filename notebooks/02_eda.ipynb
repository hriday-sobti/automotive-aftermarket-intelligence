# 02: Exploratory Data Analysis (EDA)

Audits descriptive statistics, numeric distributions, categorical mix, and inventory stockout patterns.


```python
import os, sys, pandas as pd
sales = pd.read_csv('../data/processed/fact_sales.csv')
products = pd.read_csv('../data/processed/dim_product.csv')
print(f'Total Sales Rows: {len(sales):,}')
print(f'Total Catalog SKUs: {len(products):,}')
```

```python
# Numeric summary
print(sales[['quantity', 'net_unit_price', 'revenue', 'gross_profit', 'gross_margin_pct']].describe())
```

```python
# Category revenue contribution
cat_rev = sales.groupby('sku_id').agg(revenue=('revenue', 'sum')).merge(products[['sku_id', 'product_category']], on='sku_id')
print(cat_rev.groupby('product_category')['revenue'].sum().sort_values(ascending=False))
```
