# 06: Demand Forecasting & Operational Replenishment

Evaluates out-of-sample time-series forecasting models (Naive, SMA, Exponential Smoothing) on holdout data and maps forecasts to inventory decisions.


```python
import os, sys, pandas as pd
forecasts = pd.read_csv('../data/processed/analytical_forecast_evaluation.csv')
print('Forecast Evaluation Sample:')
print(forecasts.head())
```

```python
# Compute aggregate out-of-sample holdout errors
mae_naive = (forecasts['actual_demand'] - forecasts['forecast_naive']).abs().mean()
mae_sma = (forecasts['actual_demand'] - forecasts['forecast_sma4']).abs().mean()
mae_ses = (forecasts['actual_demand'] - forecasts['forecast_exp_smooth']).abs().mean()
print(f'Naive MAE: {mae_naive:.2f}')
print(f'4-Week SMA MAE: {mae_sma:.2f}')
print(f'Exp Smoothing MAE: {mae_ses:.2f}')
```
